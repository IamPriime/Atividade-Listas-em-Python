```python
#PARTE 1
produtos = ["Camiseta Básica", "Calça Jeans", "Jaqueta Corta-Vento"]
precos = [49.90, 129.90, 219.90]

for i, j in zip(produtos, precos):
  print(f"o produto {i} custa R$ {j}")
```

```python
#PARTE 2
novo_produto = input("Digite o nome do novo produto")
novo_ preco = float(input("Digite o novo valor do produto"))

produtos.append(novo_produto)
precos.append(novo_preco)

 print(f"o produto {novo_produto} foi cadastrado com sucesso")
for i, j in zip(produtos, precos):
    print(f"o produto {i} custa R$ {j}")
```

```python
# PARTE 3 - Remover o último produto

if len(produtos) > 0:
    removido = produtos.pop()
    preco_removido = precos.pop()

    print(f"\nProduto removido: {removido}")
    print(f"Preço removido: R$ {preco_removido:.2f}")
else:
    print("Estoque vazio!")
```
